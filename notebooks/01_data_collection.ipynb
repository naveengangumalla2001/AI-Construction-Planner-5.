```python
features = [
    "plot_area_sqft",
    "builtup_area_sqft",
    "floors",
    "bedrooms",
    "bathrooms",
    "location",
    "construction_quality",
    "material_quality",
    "parking"
]

target = "total_construction_cost"

X = df[features]
y = df[target]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
numeric_features = [
    "plot_area_sqft",
    "builtup_area_sqft",
    "floors",
    "bedrooms",
    "bathrooms"
]

categorical_features = [
    "location",
    "construction_quality",
    "material_quality",
    "parking"
]
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
numeric_transformer = StandardScaler()

categorical_transformer = OneHotEncoder(
    handle_unknown="ignore"
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
```

```python
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", LinearRegression())
    ]
)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred[:10])
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
mae = mean_absolute_error(y_test, y_pred)

print("MAE:", mae)
```

```python
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print("RMSE:", rmse)
```

```python
r2 = r2_score(y_test, y_pred)

print("R² Score:", r2)
```

```python
print("Model Performance")
print("------------------")
print("MAE  :", mae)
print("RMSE :", rmse)
print("R²   :", r2)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline

rf_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", RandomForestRegressor(
            n_estimators=200,
            random_state=42,
            n_jobs=-1
        ))
    ]
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest Performance")
print("-------------------------")
print("MAE  :", rf_mae)
print("RMSE :", rf_rmse)
print("R²   :", rf_r2)
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.pipeline import Pipeline

gb_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", GradientBoostingRegressor(
            n_estimators=200,
            learning_rate=0.05,
            max_depth=3,
            random_state=42
        ))
    ]
)

gb_model.fit(X_train, y_train)

gb_pred = gb_model.predict(X_test)
```

```python
gb_mae = mean_absolute_error(y_test, gb_pred)
gb_rmse = np.sqrt(mean_squared_error(y_test, gb_pred))
gb_r2 = r2_score(y_test, gb_pred)

print("Gradient Boosting Performance")
print("-----------------------------")
print("MAE  :", gb_mae)
print("RMSE :", gb_rmse)
print("R²   :", gb_r2)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    "regressor__n_estimators": [100, 200, 300],
    "regressor__learning_rate": [0.03, 0.05, 0.1],
    "regressor__max_depth": [2, 3, 4],
    "regressor__min_samples_split": [2, 5],
    "regressor__min_samples_leaf": [1, 2]
}
```

```python
grid_search = GridSearchCV(
    estimator=gb_model,
    param_grid=param_grid,
    cv=5,
    scoring="neg_mean_absolute_error",
    n_jobs=-1,
    verbose=1
)
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:")
print(grid_search.best_params_)
```

```python
best_gb_model = grid_search.best_estimator_
```

```python
tuned_pred = best_gb_model.predict(X_test)
```

```python
tuned_mae = mean_absolute_error(y_test, tuned_pred)
tuned_rmse = np.sqrt(mean_squared_error(y_test, tuned_pred))
tuned_r2 = r2_score(y_test, tuned_pred)

print("Tuned Gradient Boosting Performance")
print("------------------------------------")
print("MAE  :", tuned_mae)
print("RMSE :", tuned_rmse)
print("R²   :", tuned_r2)
```

```python
# Predictions on training data
train_pred = best_gb_model.predict(X_train)

# Predictions on test data
test_pred = best_gb_model.predict(X_test)

# Training metrics
train_mae = mean_absolute_error(y_train, train_pred)
train_rmse = np.sqrt(mean_squared_error(y_train, train_pred))
train_r2 = r2_score(y_train, train_pred)

# Test metrics
test_mae = mean_absolute_error(y_test, test_pred)
test_rmse = np.sqrt(mean_squared_error(y_test, test_pred))
test_r2 = r2_score(y_test, test_pred)

print("Training Performance")
print("--------------------")
print("MAE  :", train_mae)
print("RMSE :", train_rmse)
print("R²   :", train_r2)

print("\nTest Performance")
print("----------------")
print("MAE  :", test_mae)
print("RMSE :", test_rmse)
print("R²   :", test_r2)
```

```python
import joblib

joblib.dump(
    best_gb_model,
    "../models/construction_cost_model.pkl"
)

print("Model saved successfully!")
```
